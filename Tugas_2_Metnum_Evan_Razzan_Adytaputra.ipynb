#Fungsi Normalisasi dan Inisialisasi data

```python
import numpy as np

#normalisasi
def min_max(data, min_val, max_val):
  data = np.array(data, dtype=float)
  if max_val - min_val == 0:
    return data

  scaled_data = (data - min_val) / (max_val - min_val)
  return scaled_data

y1 = 0.92
y2 = 0.7
y3 = 0.85
y = [y1, y2, y3]

x = np.array([[30, 9, 90],
             [20, 9, 80],
             [25, 3, 90],
             [27, 6, 85]])

data1 = x[:,0]
data1_norm = min_max(data1, 0, 40) #data min dan max sesuai yang diberikan di kelas

print(f"Data awal: {data1}")
print(f"Normalized: {data1_norm}")

data2 = x[:,1]
data2_norm = min_max(data2, 0, 15)
print(f"\nData awal: {data2}")
print(f"Normalized: {data2_norm}")

data3 = x[:,2]
data3_norm = min_max(data3, 0, 100)
print(f"\nData awal: {data3}")
print(f"Normalized: {data3_norm}")

#Setelah normalisasi:
#              a      b    c   y
#             0.75  0.6  0.9  0.92
#             0.5   0.6  0.8  0.7
#             0.625 0.2  0.9  0.85
# x_dicari :  0.675 0.4  0.85 ...

a = data1_norm[:3]
b = data2_norm[:3]
c = data3_norm[:3]
x_dicari = np.array([data1_norm[3], data2_norm[3], data3_norm[3]])

print(a)
print(b)
print(c)
print(x_dicari)
```

#Gradient Descent

```python
#asumsi
max_it = 1000
weight = np.array([0, 0, 0])
bias = 0
alpha = 0.01 #learning rate

for iterasi in range(max_it):
  y_pred1 = a[0]*weight[0] + b[0]*weight[1] + c[0]*weight[2] + bias
  y_pred2 = a[1]*weight[0] + b[1]*weight[1] + c[1]*weight[2] + bias
  y_pred3 = a[2]*weight[0] + b[2]*weight[1] + c[2]*weight[2] + bias
  y_pred = np.array([y_pred1, y_pred2, y_pred3])
  loss = np.sum((y - y_pred) ** 2)

  dw1 = 2*((y_pred - y).dot(a))
  dw2 = 2*((y_pred - y).dot(b))
  dw3 = 2*((y_pred - y).dot(c))

  dw = np.array([dw1, dw2, dw3])
  db = 2*np.sum(y_pred - y)

  weight = weight - alpha*dw
  bias = bias - alpha*db

  if loss < 0.00001: #selesaikan iterasi jika error mendekati 0
    break

  if iterasi <= 10 or iterasi % 100 == 0:
    print(f"\nIterasi ke-{iterasi}:")
    print(f"Loss: {loss}")
    print(f"Weight: {weight}")
    print(f"Bias: {bias}")
    print(f"Prediksi y: {y_pred}")

y_pred_dicari = x_dicari.dot(weight) + bias

print(f"\nWeight Akhir: {weight}")
print(f"Bias Akhir: {bias}")
print(f"Prediksi akhir:", y_pred_dicari)
print(f"loss: ", loss)
```

#Steepest Descent

```python
#asumsi awal:
max_it = 1000
weight = np.array([0, 0, 0])
bias = 0

def get_learning_rate(a, b, c, dw, bias, y):
  dw1 = dw[0]
  dw2 = dw[1]
  dw3 = dw[2]
  j = dw1*a + dw2*b + dw3*c + db

  numerator = np.sum((weight[0]*a + weight[1]*b + weight[2]*c + bias - y) * j)
  denominator = np.sum((dw1*a + dw2*b +dw3*c +db) * j)

  if denominator != 0:
    alpha = numerator / denominator
  else:
    alpha = 0.001

  return alpha


for iterasi in range(max_it):
  y_pred1 = a[0]*weight[0] + b[0]*weight[1] + c[0]*weight[2] + bias
  y_pred2 = a[1]*weight[0] + b[1]*weight[1] + c[1]*weight[2] + bias
  y_pred3 = a[2]*weight[0] + b[2]*weight[1] + c[2]*weight[2] + bias
  y_pred = np.array([y_pred1, y_pred2, y_pred3])
  loss = np.sum((y - y_pred) ** 2)

  dw1 = 2*((y_pred - y).dot(a))
  dw2 = 2*((y_pred - y).dot(b))
  dw3 = 2*((y_pred - y).dot(c))

  dw = np.array([dw1, dw2, dw3])
  db = 2*np.sum(y_pred - y)

  alpha = get_learning_rate(a, b, c, dw, bias, y)

  weight = weight - alpha*dw
  bias = bias - alpha*db

  if loss < 0.00001:
    break

  if iterasi <= 10 or iterasi % 100 == 0:
    print(f"\nIterasi ke-{iterasi}:")
    print(f"Loss: {loss}")
    print(f"Weight: {weight}")
    print(f"Bias: {bias}")
    print(f"Prediksi y: {y_pred}")

y_pred_dicari = x_dicari.dot(weight) + bias

print(f"\nWeight Akhir: {weight}")
print(f"Bias Akhir: {bias}")
print(f"Prediksi akhir:", y_pred_dicari)
print(f"loss: ", loss)
```
